# 📊 Predicción de Carga Operativa — Backoffice Call Center Financiero Honduras
**Modelo:** XGBoost Multiclase (Bajo / Medio / Alto)  
**Horizonte:** Próximos 7 días  
**Versión:** 3.0 — Cierre de mes por días hábiles exactos

### Factores de negocio incorporados
| Factor | Lógica implementada |
|--------|---------------------|
| 📆 **Cierre de mes — último día hábil** | Pico máximo. Calculado como el último día hábil real del mes (excluye fines de semana y feriados) |
| 📆 **Penúltimo día hábil** | Pico alto |
| 📆 **Antepenúltimo día hábil** | Pico moderado |
| 📆 **Primer y segundo día hábil** | Pico de inicio de mes post-cierre |
| ✂️ **Fechas de corte de cartera** | Días 5,10,15,20,23,25,28 → pico por mora |
| 🌿 **Semana Santa** | Mar/Abr — volumen cae drásticamente |
| 🇭🇳 **Semana Morazánica** | Oct — actividad reducida |
| 🎄 **Aguinaldo (Dic)** | Mayor flujo circulante |
| 📉 **Enero** | Resaca financiera de Dic |
| 💵 **14vo salario (Jun)** | 1ra quincena: menos mora |
| 📅 **Día de semana** | Lunes alto, Viernes bajo |
| 🔔 **Pre/Post feriado** | Rebote el día siguiente al feriado |

## 1. Imports y configuración

## 2. Carga de datos y feriados

```python
# ─── CONFIGURA AQUÍ ───────────────────────────────────────────────────────────
RUTA_DATOS    = '/Users/olvinfvillatorob/MastersProjects/OpsForecast_ML/Training Data/datos_solicitudes.csv'
RUTA_FERIADOS = '/Users/olvinfvillatorob/MastersProjects/OpsForecast_ML/Training Data/feriados_honduras.csv'
# ──────────────────────────────────────────────────────────────────────────────

def cargar_archivo(ruta):
    if ruta.endswith(('.xlsx', '.xls')):
        return pd.read_excel(ruta)
    return pd.read_csv(ruta, encoding='utf-8-sig')

# Solicitudes
df_raw = cargar_archivo(RUTA_DATOS)
df_raw.columns = (
    df_raw.columns.str.strip().str.upper()
    .str.replace(' ', '_').str.replace('(','',regex=False).str.replace(')','',regex=False)
)
df_raw.rename(columns={
    'CREATED_DATE':'CREATED_DATE', 'SOLICITUD_ID':'SOLICITUD_ID',
    'GESTION_TYPE':'GESTION_TYPE', 'PRODUCT_TYPETC_OR_PR':'PRODUCT_TYPE',
    'DELIQUENCY':'DELIQUENCY', 'CYCLE':'CYCLE', 'COUNTRY':'COUNTRY',
}, inplace=True)

# Feriados
df_feriados = cargar_archivo(RUTA_FERIADOS)
df_feriados.columns = df_feriados.columns.str.strip().str.lower()
df_feriados['fecha'] = pd.to_datetime(df_feriados['fecha'])
FERIADOS_SET = set(df_feriados['fecha'].dt.date)

# Semanas de vacaciones (expansión automática desde feriados de mar/abr y oct)
ss_base = df_feriados[df_feriados['fecha'].dt.month.isin([3,4])]['fecha']
SEMANA_SANTA_SET = set()
for d in ss_base:
    for delta in range(-3, 5):
        SEMANA_SANTA_SET.add((d + pd.Timedelta(days=delta)).date())

mz_base = df_feriados[
    (df_feriados['fecha'].dt.month == 10) & (df_feriados['fecha'].dt.day <= 5)
]['fecha']
MORAZAN_SET = set()
for d in mz_base:
    for delta in range(7):
        MORAZAN_SET.add((d + pd.Timedelta(days=delta)).date())

print(f'📂 Registros cargados     : {len(df_raw):,}')
print(f'📅 Feriados               : {len(FERIADOS_SET)}')
print(f'🌿 Días semana santa      : {len(SEMANA_SANTA_SET)}')
print(f'🇭🇳 Días semana morazánica : {len(MORAZAN_SET)}')
```

## 3. Utilidades — Días hábiles exactos por mes

Esta función calcula la posición real de cada día dentro del mes, contando **solo días hábiles** (excluye sábados, domingos y feriados). Así el modelo sabe exactamente si un día es el último, penúltimo o antepenúltimo hábil.

```python
_cache_habiles = {}  # Cache global para evitar recalcular

def get_habiles_mes(anio, mes, feriados_set):
    """Retorna lista ordenada de fechas hábiles del mes dado."""
    key = (anio, mes)
    if key not in _cache_habiles:
        primer = date(anio, mes, 1)
        ultimo = date(anio + 1, 1, 1) if mes == 12 else date(anio, mes + 1, 1)
        habiles = []
        d = primer
        while d < ultimo:
            if d.weekday() < 5 and d not in feriados_set:
                habiles.append(d)
            d += timedelta(days=1)
        _cache_habiles[key] = habiles
    return _cache_habiles[key]


def calcular_posicion_habil(fecha_series, feriados_set):
    """
    Para cada fecha en la serie retorna un DataFrame con:
      rank_habil_fin        : posición desde el final (1 = último hábil del mes)
      rank_habil_inicio     : posición desde el inicio (1 = primer hábil del mes)
      total_habiles_mes     : total de días hábiles del mes
      es_ultimo_habil       : 1 si es el último día hábil del mes
      es_penultimo_habil    : 1 si es el penúltimo
      es_antepenultimo_habil: 1 si es el antepenúltimo
      es_cierre_mes_habil   : 1 si está en los últimos 3 días hábiles
      es_primer_habil       : 1 si es el primer día hábil del mes
      es_segundo_habil      : 1 si es el segundo día hábil del mes
      peso_cierre           : 0→1 continuo (1 = último hábil, 0.67 = penúltimo, 0.33 = antepenúltimo)
    """
    rows = []
    for fecha in fecha_series:
        d = fecha.date() if hasattr(fecha, 'date') else fecha
        habiles = get_habiles_mes(d.year, d.month, feriados_set)
        total   = len(habiles)

        if d not in habiles:
            rows.append(dict(
                rank_habil_fin=99, rank_habil_inicio=99, total_habiles_mes=total,
                es_ultimo_habil=0, es_penultimo_habil=0, es_antepenultimo_habil=0,
                es_cierre_mes_habil=0, es_primer_habil=0, es_segundo_habil=0,
                peso_cierre=0.0,
            ))
            continue

        idx_ini = habiles.index(d)           # 0-based
        pos_ini = idx_ini + 1                # 1-based desde inicio
        pos_fin = total - idx_ini            # 1-based desde fin

        peso = max(0.0, (4 - pos_fin) / 3) if pos_fin <= 3 else 0.0

        rows.append(dict(
            rank_habil_fin         = pos_fin,
            rank_habil_inicio      = pos_ini,
            total_habiles_mes      = total,
            es_ultimo_habil        = int(pos_fin == 1),
            es_penultimo_habil     = int(pos_fin == 2),
            es_antepenultimo_habil = int(pos_fin == 3),
            es_cierre_mes_habil    = int(pos_fin <= 3),
            es_primer_habil        = int(pos_ini == 1),
            es_segundo_habil       = int(pos_ini == 2),
            peso_cierre            = round(peso, 3),
        ))
    return pd.DataFrame(rows)


# ── Verificación rápida ───────────────────────────────────────────────────────
fechas_test = pd.date_range('2024-12-20', '2024-12-31')
pos_test    = calcular_posicion_habil(fechas_test, FERIADOS_SET)
df_test     = pd.concat([pd.DataFrame({'fecha': fechas_test}), pos_test], axis=1)

print('✅ Verificación: últimos días de diciembre 2024')
print('   (25 dic = Navidad feriado | 28-29 dic = fin de semana)')
print()
print(df_test[['fecha','rank_habil_fin','es_ultimo_habil',
               'es_penultimo_habil','es_antepenultimo_habil',
               'es_cierre_mes_habil','peso_cierre']].to_string(index=False))
```

## 4. Limpieza y agregación diaria

```python
df_raw['CREATED_DATE'] = pd.to_datetime(df_raw['CREATED_DATE'], errors='coerce')
df_raw['fecha'] = df_raw['CREATED_DATE'].dt.normalize()
df_raw.dropna(subset=['fecha'], inplace=True)

if df_raw['COUNTRY'].nunique() > 1:
    df_raw = df_raw[df_raw['COUNTRY'].str.upper().str.contains('HND|HONDURAS', na=False)]

daily = (
    df_raw.groupby('fecha')
    .agg(
        total_solicitudes = ('SOLICITUD_ID',  'count'),
        pct_TC            = ('PRODUCT_TYPE',  lambda x: (x.str.upper()=='TC').mean()),
        avg_deliquency    = ('DELIQUENCY',    lambda x: pd.to_numeric(x, errors='coerce').mean()),
        avg_cycle         = ('CYCLE',         lambda x: pd.to_numeric(x, errors='coerce').mean()),
        n_gestiones_tipo  = ('GESTION_TYPE',  'nunique'),
        pct_cobro         = ('GESTION_TYPE',  lambda x: (x=='COBRO').mean()),
        pct_acuerdo       = ('GESTION_TYPE',  lambda x: (x=='ACUERDO_PAGO').mean()),
        pct_pago_parcial  = ('GESTION_TYPE',  lambda x: (x=='PAGO_PARCIAL').mean()),
    )
    .reset_index().sort_values('fecha')
)

rango = pd.date_range(daily['fecha'].min(), daily['fecha'].max(), freq='D')
daily = daily.set_index('fecha').reindex(rango).rename_axis('fecha').reset_index()
daily['total_solicitudes'].fillna(0, inplace=True)

print(f'📊 Rango  : {daily["fecha"].min().date()} → {daily["fecha"].max().date()}')
print(f'📅 Días   : {len(daily):,} totales')
```

## 5. Feature Engineering

```python
def build_features(df, feriados_set, semana_santa_set, morazan_set, fechas_corte):
    d = df.copy()
    f = d['fecha']

    # ── 1. Temporales básicas ─────────────────────────────────────────────────
    d['dia_semana']    = f.dt.dayofweek
    d['dia_mes']       = f.dt.day
    d['mes']           = f.dt.month
    d['trimestre']     = f.dt.quarter
    d['anio']          = f.dt.year
    d['semana_anio']   = f.dt.isocalendar().week.astype(int)
    d['es_fin_semana'] = (d['dia_semana'] >= 5).astype(int)
    d['es_lunes']      = (d['dia_semana'] == 0).astype(int)
    d['es_viernes']    = (d['dia_semana'] == 4).astype(int)

    # Señales cíclicas (capturan periodicidad sin corte artificial)
    d['sin_dia_mes']   = np.sin(2 * np.pi * d['dia_mes'] / 31)
    d['cos_dia_mes']   = np.cos(2 * np.pi * d['dia_mes'] / 31)
    d['sin_mes']       = np.sin(2 * np.pi * d['mes'] / 12)
    d['cos_mes']       = np.cos(2 * np.pi * d['mes'] / 12)
    d['sin_diasem']    = np.sin(2 * np.pi * d['dia_semana'] / 7)
    d['cos_diasem']    = np.cos(2 * np.pi * d['dia_semana'] / 7)

    # ── 2. Cierre de mes — días hábiles exactos ───────────────────────────────
    pos_df = calcular_posicion_habil(f, feriados_set)
    for col in pos_df.columns:
        d[col] = pos_df[col].values

    # ── 3. Feriados ───────────────────────────────────────────────────────────
    d['es_feriado']   = f.dt.date.isin(feriados_set).astype(int)
    d['pre_feriado']  = f.apply(lambda x: int((x + pd.Timedelta(days=1)).date() in feriados_set))
    d['post_feriado'] = f.apply(lambda x: int((x - pd.Timedelta(days=1)).date() in feriados_set))

    # ── 4. Semana Santa ───────────────────────────────────────────────────────
    d['es_semana_santa']  = f.dt.date.isin(semana_santa_set).astype(int)
    d['pre_semana_santa'] = f.apply(
        lambda x: int(any((x + pd.Timedelta(days=i)).date() in semana_santa_set for i in range(1,6))))

    # ── 5. Semana Morazánica ──────────────────────────────────────────────────
    d['es_semana_morazan']  = f.dt.date.isin(morazan_set).astype(int)
    d['pre_semana_morazan'] = f.apply(
        lambda x: int(any((x + pd.Timedelta(days=i)).date() in morazan_set for i in range(1,4))))

    # ── 6. Estacionalidad financiera ──────────────────────────────────────────
    d['es_enero']         = (d['mes'] == 1).astype(int)
    d['es_dic']           = (d['mes'] == 12).astype(int)
    d['es_dic_aguinaldo'] = ((d['mes'] == 12) & d['dia_mes'].between(10, 20)).astype(int)
    d['es_jun_14vo']      = ((d['mes'] == 6) & (d['dia_mes'] <= 15)).astype(int)
    d['es_jun_2da']       = ((d['mes'] == 6) & (d['dia_mes'] > 15)).astype(int)

    # ── 7. Fechas de corte de cartera ─────────────────────────────────────────
    d['dist_corte']     = d['dia_mes'].apply(lambda dm: min(abs(dm - c) for c in fechas_corte))
    d['es_dia_corte']   = d['dia_mes'].isin(fechas_corte).astype(int)
    d['es_cerca_corte'] = (d['dist_corte'] <= 2).astype(int)
    # Peso inverso a distancia al corte (0→1)
    d['peso_corte']     = (1 / (d['dist_corte'] + 1)).round(3)

    # ── 8. Lags ───────────────────────────────────────────────────────────────
    for lag in [1, 2, 3, 5, 7, 14, 21, 28]:
        d[f'lag_{lag}d'] = d['total_solicitudes'].shift(lag)

    # ── 9. Rolling statistics ─────────────────────────────────────────────────
    s = d['total_solicitudes'].shift(1)
    for w in [7, 14, 30]:
        d[f'roll_mean_{w}d'] = s.rolling(w).mean()
        d[f'roll_std_{w}d']  = s.rolling(w).std()
        d[f'roll_max_{w}d']  = s.rolling(w).max()
        d[f'roll_min_{w}d']  = s.rolling(w).min()

    # ── 10. Variables contextuales del día ────────────────────────────────────
    ctx_cols = ['pct_TC','avg_deliquency','avg_cycle','n_gestiones_tipo',
                'pct_cobro','pct_acuerdo','pct_pago_parcial']
    for col in ctx_cols:
        if col in d.columns:
            d[col] = d[col].fillna(d[col].median())

    return d


daily = build_features(daily, FERIADOS_SET, SEMANA_SANTA_SET, MORAZAN_SET, FECHAS_CORTE)

FEATURE_COLS = [c for c in daily.columns if c not in ['fecha','total_solicitudes','carga']]
print(f'✅ Features construidas: {len(FEATURE_COLS)}')

# Mostrar features de cierre de mes
cols_cierre = [c for c in FEATURE_COLS if 'habil' in c or 'cierre' in c or 'peso' in c]
print(f'\n📆 Features de cierre de mes ({len(cols_cierre)}):')
print(cols_cierre)
```

## 7. Etiquetado de carga operativa

```python
dias_activos = daily[daily['total_solicitudes'] > 0]['total_solicitudes']
UMBRAL_BAJO  = dias_activos.quantile(0.33)
UMBRAL_ALTO  = dias_activos.quantile(0.67)

print(f'Umbral BAJO  (P33): {UMBRAL_BAJO:.0f} solicitudes/día')
print(f'Umbral ALTO  (P67): {UMBRAL_ALTO:.0f} solicitudes/día')

daily['carga'] = daily['total_solicitudes'].apply(
    lambda n: 'BAJO' if n <= UMBRAL_BAJO else ('MEDIO' if n <= UMBRAL_ALTO else 'ALTO')
)
print('\nDistribución de clases:')
print(daily[daily['total_solicitudes'] > 0]['carga'].value_counts())
```

## 8. División temporal Train / Validación

```python
FEATURE_COLS  = [c for c in daily.columns if c not in ['fecha','total_solicitudes','carga']]
daily_clean   = daily.dropna(subset=FEATURE_COLS).copy()
DIAS_VAL      = 90
fecha_corte   = daily_clean['fecha'].max() - pd.Timedelta(days=DIAS_VAL)

train = daily_clean[daily_clean['fecha'] <= fecha_corte].copy()
val   = daily_clean[daily_clean['fecha'] >  fecha_corte].copy()

le = LabelEncoder()
le.fit(['BAJO','MEDIO','ALTO'])

X_train, y_train = train[FEATURE_COLS], le.transform(train['carga'])
X_val,   y_val   = val[FEATURE_COLS],   le.transform(val['carga'])

print(f'Train  : {len(train):,} días  ({train["fecha"].min().date()} → {train["fecha"].max().date()})')
print(f'Val    : {len(val):,} días  ({val["fecha"].min().date()} → {val["fecha"].max().date()})')
print(f'Features: {len(FEATURE_COLS)}')
```

## 9. Entrenamiento XGBoost

```python
modelo = XGBClassifier(
    objective             = 'multi:softprob',
    num_class             = 3,
    n_estimators          = 600,
    learning_rate         = 0.04,
    max_depth             = 5,
    min_child_weight      = 3,
    subsample             = 0.80,
    colsample_bytree      = 0.75,
    gamma                 = 0.1,
    reg_alpha             = 0.1,
    reg_lambda            = 1.0,
    use_label_encoder     = False,
    eval_metric           = 'mlogloss',
    early_stopping_rounds = 40,
    random_state          = RANDOM_STATE,
    verbosity             = 0,
)

modelo.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=100)
print(f'\n✅ Mejor iteración : {modelo.best_iteration}')
print(f'   Log-loss val    : {modelo.best_score:.4f}')
```

## 10. Evaluación

```python
y_pred        = modelo.predict(X_val)
y_pred_labels = le.inverse_transform(y_pred)
y_val_labels  = le.inverse_transform(y_val)

print(f'🎯 Accuracy: {accuracy_score(y_val, y_pred):.2%}\n')
print(classification_report(y_val_labels, y_pred_labels, target_names=['ALTO','BAJO','MEDIO']))
```

## 11. Predicción de los próximos 7 días

```python
def predecir_proximos_dias(modelo, daily_clean, feriados_set, semana_santa_set,
                            morazan_set, fechas_corte, le, feature_cols, n_dias=7):
    media_nivel = {
        nivel: daily_clean[daily_clean['carga']==nivel]['total_solicitudes'].mean()
        for nivel in ['BAJO','MEDIO','ALTO']
    }
    COLS_CTX = ['fecha','total_solicitudes','pct_TC','avg_deliquency','avg_cycle',
                'n_gestiones_tipo','pct_cobro','pct_acuerdo','pct_pago_parcial']
    contexto = daily_clean[[c for c in COLS_CTX if c in daily_clean.columns]].copy()

    ultima = contexto['fecha'].max()
    resultados = []

    for i in range(1, n_dias + 1):
        fp = ultima + pd.Timedelta(days=i)
        nueva = {'fecha': fp, 'total_solicitudes': np.nan,
                 'pct_TC': contexto['pct_TC'].iloc[-30:].mean(),
                 'avg_deliquency': contexto['avg_deliquency'].iloc[-30:].mean(),
                 'avg_cycle': contexto['avg_cycle'].iloc[-30:].mean(),
                 'n_gestiones_tipo': contexto['n_gestiones_tipo'].iloc[-14:].mean(),
                 'pct_cobro': contexto['pct_cobro'].iloc[-14:].mean(),
                 'pct_acuerdo': contexto['pct_acuerdo'].iloc[-14:].mean(),
                 'pct_pago_parcial': contexto['pct_pago_parcial'].iloc[-14:].mean()}
        contexto = pd.concat([contexto, pd.DataFrame([nueva])], ignore_index=True)

        ctx_f = build_features(contexto, feriados_set, semana_santa_set, morazan_set, fechas_corte)
        fila  = ctx_f[ctx_f['fecha'] == fp]
        if fila.empty:
            resultados.append({'fecha': fp, 'pred_carga': 'SIN DATOS'})
            continue

        X      = fila[feature_cols].fillna(method='ffill')
        proba  = modelo.predict_proba(X)[0]
        idx    = np.argmax(proba)
        pred   = le.inverse_transform([idx])[0]

        contexto.loc[contexto['fecha']==fp, 'total_solicitudes'] = media_nivel[pred]

        # Detectar contexto del día para explicación
        fp_date = fp.date()
        pos     = calcular_posicion_habil(pd.Series([fp]), feriados_set).iloc[0]
        contexto_dia = []
        if pos['es_ultimo_habil']:        contexto_dia.append('🔴 CIERRE MES (último hábil)')
        elif pos['es_penultimo_habil']:   contexto_dia.append('🟠 Penúltimo hábil')
        elif pos['es_antepenultimo_habil']:contexto_dia.append('🟡 Antepenúltimo hábil')
        if pos['es_primer_habil']:        contexto_dia.append('📋 Inicio de mes')
        if fp.month == 12:                contexto_dia.append('🎄 Aguinaldo')
        if fp.month == 1:                 contexto_dia.append('📉 Enero (bajo flujo)')
        if fp_date in feriados_set:       contexto_dia.append('🔕 Feriado')
        if fp_date in semana_santa_set:   contexto_dia.append('🌿 Semana Santa')
        if fp_date in morazan_set:        contexto_dia.append('🇭🇳 Semana Morazánica')
        if fp.day in fechas_corte:        contexto_dia.append(f'✂️ Fecha de corte (día {fp.day})')

        resultados.append({
            'fecha'       : fp,
            'dia'         : fp.strftime('%A'),
            'pred_carga'  : pred,
            'prob_bajo'   : round(proba[le.transform(['BAJO'])[0]], 3),
            'prob_medio'  : round(proba[le.transform(['MEDIO'])[0]], 3),
            'prob_alto'   : round(proba[le.transform(['ALTO'])[0]], 3),
            'rank_habil_fin': int(pos['rank_habil_fin']),
            'contexto'    : ' | '.join(contexto_dia) if contexto_dia else 'Día normal',
        })

    return pd.DataFrame(resultados)


df_pred = predecir_proximos_dias(
    modelo, daily_clean, FERIADOS_SET, SEMANA_SANTA_SET,
    MORAZAN_SET, FECHAS_CORTE, le, FEATURE_COLS, n_dias=7
)

display(df_pred[['fecha','dia','pred_carga','prob_bajo','prob_medio','prob_alto',
                 'rank_habil_fin','contexto']])
```

```python
COLORES = {'BAJO':'#2ecc71','MEDIO':'#f39c12','ALTO':'#e74c3c','SIN DATOS':'#bdc3c7'}

fig, ax = plt.subplots(figsize=(14, 6))

x = range(len(df_pred))
ax.bar(x, df_pred['prob_alto'],  color='#e74c3c', label='P(ALTO)',  alpha=0.85)
ax.bar(x, df_pred['prob_medio'], bottom=df_pred['prob_alto'],
       color='#f39c12', label='P(MEDIO)', alpha=0.75)
ax.bar(x, df_pred['prob_bajo'],  bottom=df_pred['prob_alto']+df_pred['prob_medio'],
       color='#2ecc71', label='P(BAJO)',  alpha=0.65)

etiq_x = []
for _, row in df_pred.iterrows():
    if 'dia' not in row or pd.isna(row.get('dia')): 
        etiq_x.append('?')
        continue
    linea = row['dia'][:3] + '\n' + str(row['fecha'].day)
    ctx   = row.get('contexto','')
    if 'último hábil' in ctx.lower():  linea += '\n🔴CIERRE'
    elif 'Penúltimo'   in ctx:         linea += '\n🟠pen'
    elif 'Antepenúltimo' in ctx:       linea += '\n🟡ante'
    elif 'Inicio'      in ctx:         linea += '\n📋inicio'
    elif 'Feriado'     in ctx:         linea += '\n🔕feriado'
    etiq_x.append(linea)

for i, (_, row) in enumerate(df_pred.iterrows()):
    c = COLORES.get(row.get('pred_carga','SIN DATOS'), '#555')
    ax.text(i, 1.03, row.get('pred_carga','?'),
            ha='center', va='bottom', fontsize=10, fontweight='bold', color=c)

ax.set_xticks(list(x))
ax.set_xticklabels(etiq_x, fontsize=9)
ax.set_ylim(0, 1.22)
ax.set_ylabel('Probabilidad acumulada')
ax.set_title('📅 Pronóstico de carga operativa — Próximos 7 días', fontsize=13, pad=12)
ax.legend(loc='upper right', fontsize=9)
ax.axhline(0.5, linestyle='--', color='gray', alpha=0.4, linewidth=0.8)
plt.tight_layout()
plt.show()
```
